```python
import re
import pandas as pd  # Assuming pandas is already imported, but just in case

##MATCH PICK 포함!!!!

def categorizeBigMP(df, category, category_keywords=category_keywords):
    # 카테고리별 pick_list와 remove_list 정의

    # 유효한 카테고리인지 확인
    if category not in category_keywords:
        raise ValueError(f"지원하지 않는 카테고리: {category}. 지원 카테고리: {list(category_keywords.keys())}")
    
    # pick_list와 remove_list 가져오기
    pick_list = category_keywords[category]['pick_list']
    remove_list = category_keywords[category]['remove_list']
    
    # 정규표현식 패턴 생성
    pick_pattern = '|'.join(pick_list)
    
    # pick_list로 필터링
    result_df = df[df['입찰명'].str.contains(pick_pattern, na=False, regex=True)].copy()  # copy to avoid SettingWithCopyWarning
    
    # matched_pick_keywords 열 추가: 각 행에서 매치된 pick 키워드들 (unique list as string)
    def find_matched_keywords(text, keywords):
        matched = set()
        for kw in keywords:
            if re.search(kw, text):
                matched.add(kw)
        return ', '.join(sorted(matched)) if matched else ''
    
    result_df[f'{category}_Pick'] = result_df['입찰명'].apply(lambda x: find_matched_keywords(x, pick_list))
    
    # remove_list가 비어 있지 않으면 제거 필터링
    if remove_list and remove_list != ['']:
        remove_pattern = '|'.join(remove_list)
        
        # matched_remove_keywords 열 추가 (필터링 전에 추가, 하지만 결과에서는 매치된 행이 제외되므로 대부분 빈 문자열)
        result_df[f'{category}_Remove'] = result_df['입찰명'].apply(lambda x: find_matched_keywords(x, remove_list))
        
        # remove_pattern에 매치되는 행 제외
        result_df = result_df[~result_df['입찰명'].str.contains(remove_pattern, na=False, regex=True)]
    else:
        # remove_list가 없으면 Matched_Remove 빈 문자열
        result_df[f'{category}_Remove'] = ''
    
    # 필요한 컬럼만 선택 (기존 컬럼 + 새로운 두 열, Matched_Remove를 두 번째 새로운 열로, 하지만 순서상 마지막에)
    # result_df = result_df[['입찰명', '발주기관', '낙찰자(상호)', f'{category}_Pick', f'{category}_Remove']]
    # print(f'{category} 결과 건수: {len(result_df)}건')

    return result_df
```

```python
# 중복 데이터를 수집할 리스트
data = []

# 외부 루프: 1종부터 10종까지
for category in range(1, 11):
    # 내부 루프: 중복 쌍을 피하기 위해 category보다 큰 i만 비교 (대칭이므로)
    for i in range(category + 1, 11):
        # checkDuplicatesInBidName 함수 호출 (사용자가 이미 정의했다고 가정)
        a = checkDuplicatesInBidName(f'{category}종', f'{i}종')
        duplicates = len(a)
        
        # 데이터 append (cat1을 작은 번호로 유지)
        data.append({
            'cat1': f'{category}종',
            'cat2': f'{i}종',
            'duplicates': duplicates
        })

# DataFrame 생성 (간단한 테이블 형식)
df = pd.DataFrame(data)

# 결과 출력 (보기 쉽게)
print("중복 카운트 테이블 (긴 형식):")
print(df)

# 피벗 테이블로 변환 (행렬 형식, 대칭 행렬로 만들기)
# 10x10 빈 행렬 생성
categories = [f'{i}종' for i in range(1, 11)]
matrix = pd.DataFrame(0, index=categories, columns=categories)

# data에서 값을 채우고 대칭으로 복사
for row in data:
    cat1 = row['cat1']
    cat2 = row['cat2']
    dup = row['duplicates']
    matrix.loc[cat1, cat2] = dup
    matrix.loc[cat2, cat1] = dup  # 대칭

# 대각선은 0으로 유지 (자기 자신 비교 아님)

# 피벗 결과 출력
print("\n중복 카운트 행렬 (피벗 형식):")
print(matrix)

# 추가 팁: Jupyter나 Colab에서 실행하면 아래처럼 히트맵으로 시각화 가능
# display(matrix.style.background_gradient(cmap='viridis', axis=None))
```
