```python
import pandas as pd
import streamlit as st

# CSV 불러오기 (경로 꼭 확인하세요)
df = pd.read_csv("C:/Users/송하선/수요기관_전체분류_보강분류.csv", encoding="cp949")

st.set_page_config(layout="wide")
st.title("군수품 관리 대시보드")

# --- 사이드바 필터 ---
st.sidebar.header("조회 조건")

# 군수품 분류 전체 목록
if "군수품분류" in df.columns:
    all_categories = ["전체"] + sorted(df["군수품분류"].dropna().unique())
else:
    st.error("❌ CSV에 '군수품분류' 컬럼이 없습니다. 컬럼명을 확인해주세요.")
    st.stop()

물자분류 = st.sidebar.selectbox("대분류", all_categories)

군구분목록 = ["전체", "육군", "해군", "공군", "국방부직할부대"]
군구분 = st.sidebar.selectbox("군 구분", 군구분목록)

검색어 = st.sidebar.text_input("검색어 입력 (입찰공고명)")

# --- 필터 적용 ---
filtered_df = df.copy()
if 물자분류 != "전체":
    filtered_df = filtered_df[filtered_df["군수품분류"] == 물자분류]
if 군구분 != "전체" and "수요기관명" in df.columns:
    filtered_df = filtered_df[filtered_df["수요기관명"].str.contains(군구분, na=False)]
if 검색어 and "입찰공고명" in df.columns:
    filtered_df = filtered_df[filtered_df["입찰공고명"].str.contains(검색어, na=False)]

# --- 메인 화면 ---
col1, col2 = st.columns([2, 1])

with col1:
    st.subheader("조회 결과")
    st.write("PLZ ATTACH IMG 1.")
    st.dataframe(filtered_df, width=1200)

with col2:
    st.subheader("상세 정보")
    st.write("PLZ ATTACH IMG 2.")
    if not filtered_df.empty:
        선택행 = st.selectbox("확인할 행 선택", filtered_df.index)
        st.json(filtered_df.loc[선택행].to_dict())
    else:
        st.info("조건에 맞는 데이터가 없습니다.")

# --- 하단 통계 요약 ---
st.subheader("통계 요약")
st.write("PLZ ATTACH IMG 3.")
summary = filtered_df["군수품분류"].value_counts()
st.bar_chart(summary)
```
