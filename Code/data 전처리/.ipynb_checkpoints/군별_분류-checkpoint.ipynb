```python
from bs4 import BeautifulSoup
import requests
import numpy as np
import pandas as pd
import json
from urllib.request import urlopen

df = pd.read_csv("C:/theBoram4Jo/hasun/Total 물품 낙찰 결과.csv", encoding="cp949")

if "발주기관" in df.columns:
    print("발주기관")
    print(df["발주기관"])  # 앞 20개만
    print("총 개수:", df["발주기관"].notna().sum())
else:
    print("⚠️ '발주기관' 컬럼이 없습니다. 실제 컬럼들:", df.columns.tolist())
```

```python
all_orgs = df["발주기관"]

unique_orgs = all_orgs.drop_duplicates()
print(df["발주기관"].drop_duplicates().to_string(index=False))
print("고유 기관 개수:", unique_orgs.nunique())

unique_list = unique_orgs.tolist()
```

```python
unique_orgs = df["발주기관"].drop_duplicates()

army = unique_orgs[unique_orgs.str.contains("육군", na=False)]
navy = unique_orgs[unique_orgs.str.contains("해군", na=False)]
air  = unique_orgs[unique_orgs.str.contains("공군", na=False)]
mask = ~(unique_orgs.str.contains("육군|해군|공군", na=False))
others = unique_orgs[mask]

print("육군")
print(army.to_string(index=False))

print("\n\n해군")
print(navy.to_string(index=False))

print("\n\n공군")
print(air.to_string(index=False))

print("\n\n기타")
print(others.to_string(index=False))
```

```python
df["군 분류"] = "기타"

# 조건별로 갱신
df.loc[df["발주기관"].str.contains("육군", na=False), "군 분류"] = "육군"
df.loc[df["발주기관"].str.contains("해군", na=False), "군 분류"] = "해군"
df.loc[df["발주기관"].str.contains("공군", na=False), "군 분류"] = "공군"
```

```python
df.to_csv(r"C:/theBoram4Jo/hasun/Total 물품 낙찰 결과.csv", 
          index=False, 
          encoding="utf-8-sig")
```

```python
all_keywords = {
    "2종-피복류": [
        "군복","피복","방한","침구","장구","의류","방탄복","군화","모자","양말","티셔츠","속옷","안전화","작업복","전투복","일용품","특수임무피복",
        "군장류","안전모","장갑","비옷","보호복"
    ],
    "2종-병참물자": [
        "비품","기재","취사","냉난방","군악","체육","교육","장구","포장재","실험기구","안전용품","로프","비품류","장비류","기구류",
        "베어링","볼트","너트","와셔","패킹","씰","가스켓","체인","기어","필터","패드","브러시","벨트","휠","스프링","법무","군사경찰",
        "인쇄","군종","공보정훈","항공장구","잠수장구","기류","해상표적",
        "정훈자료","군악기","교육훈련자재","함정전용품","산업안전용품"
    ],
    "2종-화생방물자": [
        "탐지","보호장비","방독면","치료물자","연막","제독제","구조기구","오염표지판","지역장비","독성제거제","작도기재",
        "개인제독제","제독장비","화학탐지기","화학방호복"
    ],
    "2종-통신물자": [
        "전지","배터리","야전선","CCTV","무전기","전화","팩스","라디오","위성TV","전산","네트워크","컴퓨터","휴대용무전기","수신기",
        "군용전화기","교환기","전술통신","네트워크장비"
    ],
    "2종-공병물자": [
        "소방","소화기","위장망","소방피복","소방장구",
        "철조망","방책자재","위장자재"
    ],
    "2종-기타": [
        "군사지도","지도","드론","육도","해도","항공도","소형드론","2kg미만드론"
    ],

    "3종-일반유류": ["경유","휘발유","등유"],
    "3종-항공류": ["제트유","항공유","항공휘발유","AV-GAS"],
    "3종-윤활유": ["엔진오일","기어오일","유압작동오일","압축기오일","그리스","방청유","절삭유","절연유"],
    "3종-화공약품류": ["방청제","세관제","솔벤트","부동액","제초제","정수약품","방역약품","살충제",
        "세정제","소독약","염산","황산","가성소다"
    ],
    "3종-고체연료": ["목재펠릿","연탄","공탄","조개탄","착화탄","고형연료"],
    "3종-가스류": ["LPG","LNG","CNG","아세틸렌","냉매","가스","산소","질소","아르곤"],
    "3종-유류포장재료": ["드럼","드럼통","가스용기"],

    "5종-지상탄약": [
        "소구경탄","직사화기탄","박격포탄","포병탄","지대지유도탄","수류탄","지뢰","연막탄","신호탄","화학탄","폭약","폭파기재",
        "축사기탄","교보재용연막탄"
    ],
    "5종-해상탄약": [
        "함포탄","수중탄","기뢰","폭뢰","함정유도탄","RBOC","DAGAIE","신관","교보재용탄약"
    ],
    "5종-항공탄약": [
        "항공유도탄","투하탄","일반폭탄","확산탄","기관포탄","항공로켓탄","심리전탄","전자전탄","자폭드론"
    ],
    "5종-기타": ["탄피","탄약포장","비축원자재"]
}

def classify_from_pdf(text):
    if pd.isna(text):
        return None
    s = str(text)
    for category, kws in all_keywords.items():
        for kw in kws:
            if kw in s:
                return category
    return None

# 입찰공고명 기준 분류
df["군수품분류"] = df["입찰명"].apply(classify_from_pdf)
```

```python
df = pd.read_csv(r"C:\theBoram4Jo\hasun\Total 물품 낙찰 결과_군_분류_군수품_2종4종분류_추가.csv", encoding="cp949")

rules = {
    '4종': {
        'pick_list': ['목재류','건축용', '아파트', '숙소', '제조설치', '철재', '철근', '철선',
                      '페인트', '시멘트', '보일러', '콘크리트', '아스팔트', '벽돌', '타일', '생활관',
                      '토목', '강관', '신축'],
        'remove_list': ['병원', '의무', '치료']
    },
    '8종': {
        'pick_list': ['병원', '환자', '의무', '외상', '안과', '자극', '신경', '의학',
                      '이비인후과', '치과', '붕대', '외과', '수술', '의료', '병리', '예방약',
                      '방사선장비', '의약품', '치료', '안경제작', '위생'],
        'remove_list': ['신축', '보일러', '제조설치']
    }
}

# 기준이 되는 텍스트 컬럼 (예: 입찰공고명)
target_col = "입찰명"

# 규칙 적용
for 종, rule in rules.items():
    pick_list = rule["pick_list"]
    remove_list = rule["remove_list"]

    # pick 조건
    mask_pick = df[target_col].str.contains("|".join(pick_list), na=False)

    # remove 조건
    mask_remove = df[target_col].str.contains("|".join(remove_list), na=False)

    # 최종 적용: pick O & remove X 인 행만 해당 종으로 지정
    df.loc[mask_pick & ~mask_remove, "군수품분류"] = 종
```
