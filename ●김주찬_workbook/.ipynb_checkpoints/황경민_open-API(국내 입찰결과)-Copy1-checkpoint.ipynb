#### open API 불러오기

```python
base_url = 'http://openapi.d2b.go.kr/openapi/service/BidResultInfoService'
service_name = 'getDmstcSuccessBidResult'
serviceKey = '8bea1c13542aaa1626e136b275cde504cb37de3c3a31d19cd15d7759b330efdf'
opengDateBegin  = 20160101         #개찰일자시작
opengDateEnd    = 20161231         #개찰일자종료
# bidNm=	                       #입찰명 
# orntCode=	                       #발주기관코드 
# g2bPblancNoOdr=	               #G2B공고번호차수
numOfRows       = 10000            #페이지당항목수
pageNo          = 1	               #페이지



params = f'serviceKey={serviceKey}&opengDateBegin={opengDateBegin}&opengDateEnd={opengDateEnd}&numOfRows={numOfRows}&pageNo={pageNo}'

url = f'{base_url}{service_name}?{params}'
url
```

#### BeautifulSoup을 이용하여 XML 자료 가져오기

```python
xml = requests.get(f'{base_url}{service_name}', params=params).text
soup = BeautifulSoup(xml, 'xml')
items = soup.find_all('item')
len(items)
```

```python
tag_names = ['bidCnt','pblancNo','pblancOdr','dcsNo','iemNo','bidNm','busiDivs','orntCode','ornt','opengDt','tbidCode','tbidName','tbidEname','tbidRptr','tbidTel','tbidAddr','tbidRate','tbidAmount']
col_names = ['참가수', '공고번호', '공고차수', '판단번호', '항목번호', '입찰명', '업무구분', '발주기관코드', '발주기관', '개찰일시', '낙찰자(업체코드)', '낙찰자(상호)', '낙찰자(영문상호)', '낙찰자(대표자)', '낙찰자(연락처)', '낙찰자(주소)', '낙찰률', '낙찰금액']
list(zip(tag_names, col_names))
```

```python
def to_extract_row(item, tag_names, col_names=None):
    data = {}
    if col_names is None:
        col_names = tag_names
    for tag, col in zip(tag_names, col_names):
        tagobj = item.find(tag)
        data[col] = tagobj.text.strip() if tagobj else None
    return data
```

```python
item_list = [to_extract_row(item_tag, tag_names, col_names) for item_tag in soup.find_all('item')]
```

```python
df= pd.DataFrame(item_list)
df.info()
```

# 자동함수 만들기

```python
def request_estate_data(opengDateBegin  = '20180101',
                        opengDateEnd    = '20181231',
                        # bidNm=	               
                        # orntCode=	               
                        # g2bPblancNoOdr=	       
                        numOfRows       = '15000', 
                        pageNo          = '1'	       ):
    serviceKey = '8bea1c13542aaa1626e136b275cde504cb37de3c3a31d19cd15d7759b330efdf'
    base_url = 'http://openapi.d2b.go.kr/openapi/service/BidResultInfoService'
    service_name = 'getDmstcSuccessBidResult'
    params = 'opengDateBegin=' + opengDateBegin + '&' \
           + 'opengDateEnd=' + opengDateEnd + '&' \
           + 'serviceKey=' + serviceKey + '&' \
           + 'pageNo=' + pageNo + '&' \
           + 'numOfRows=' + numOfRows
    url = f'{base_url}{service_name}?{params}'
    
    xml = requests.get(url).text
    soup = BeautifulSoup(xml, 'xml')
    return soup

tag_names = ['bidCnt','pblancNo','pblancOdr','dcsNo','iemNo','bidNm','busiDivs','orntCode','ornt','opengDt','tbidCode','tbidName','tbidEname','tbidRptr','tbidTel','tbidAddr','tbidRate','tbidAmount']
col_names = ['참가수', '공고번호', '공고차수', '판단번호', '항목번호', '입찰명', '업무구분', '발주기관코드', '발주기관', '개찰일시', '낙찰자(업체코드)', '낙찰자(상호)', '낙찰자(영문상호)', '낙찰자(대표자)', '낙찰자(연락처)', '낙찰자(주소)', '낙찰률', '낙찰금액']

def extract_data_and_DF(soup, tag_names, col_names):
    item_list = [to_extract_row(item_tag, tag_names, col_names) for item_tag in soup.find_all('item')]
    return pd.DataFrame(item_list)
```

```python
soup = request_estate_data()
df3 = extract_data_and_DF(soup, tag_names, col_names)
df3
```
