```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
import os
import wordcloud
from wordcloud import WordCloud
from konlpy.tag import Okt, Kkma, Komoran
from collections import Counter
import pickle


```

```python
result_pr = pd.read_csv("C:\crawling_project\data\입찰결과물품_df.csv", encoding='cp949') 
```

```python
# result_pr['입찰공고명'][90:100]
```

```python
result_name = result_pr['입찰공고명'].tolist()  # ✅ 리스트[str]로 변환
result_name[0]
```

```python

okt = Okt()
tokens_noun =[]
for text in result_name:
    tokens = okt.nouns(text)
    tokens_noun.append(tokens)
flattened_tokens = [token for sublist in tokens_noun for token in sublist]
tokens2 = [token for token in flattened_tokens if len(token)>1]
```

```python
len(tokens2)
```

```python

## 토큰화 & 명사들 & 2글자이상만 토큰별 빈도수
from collections import Counter

token_cnt = Counter(tokens2)
token_cnt_df = pd.DataFrame(data=pd.Series(token_cnt), columns=['Freq'])
token_cnt_df.index.name = 'Word'
token_cnt_df
```

```python
token_cnt_df = token_cnt_df.sort_values(by='Freq', ascending=False)

# 2. 5개 이상 개수 있는 행만 필터링
token_cnt_df = token_cnt_df[token_cnt_df['Freq'] >= 5]

# token_cnt_df.to_csv('frequency.csv', encoding='cp949')
```
