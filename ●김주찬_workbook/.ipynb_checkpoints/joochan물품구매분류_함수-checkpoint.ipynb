```python
def categorizeBig(df, category, category_keywords=category_keywords):
    # 카테고리별 pick_list와 remove_list 정의

    # 유효한 카테고리인지 확인
    if category not in category_keywords:
        raise ValueError(f"지원하지 않는 카테고리: {category}. 지원 카테고리: {list(category_keywords.keys())}")
    
    # pick_list와 remove_list 가져오기
    pick_list = category_keywords[category]['pick_list']
    remove_list = category_keywords[category]['remove_list']
    
    # 정규표현식 패턴 생성
    pick_pattern = '|'.join(pick_list)
    
    # pick_list로 필터링
    result_df = PBL_df[PBL_df['입찰명'].str.contains(pick_pattern, na=False, regex=True)]
    
    # remove_list가 비어 있지 않으면 제거 필터링
    if remove_list and remove_list != ['']:
        remove_pattern = '|'.join(remove_list)
        result_df = result_df[~result_df['입찰명'].str.contains(remove_pattern, na=False, regex=True)]
    
    # 필요한 컬럼만 선택
    result_df = result_df[['입찰명', '발주기관', '낙찰자(상호)']]
    print(f'{category} 결과 건수: {len(result_df)}건')

    return result_df


len(categorizeBig(PBL_df, '2종'))
# 일종df = categorizeBig(입찰결과물품df, '1종')
# output_path = os.path.join(folder_path, '일종DF.csv')
# 일종df.to_csv(output_path, encoding='cp949')

# 테스트df = categorizeBig(입찰결과물품df, '테스트')
# output_path = os.path.join(folder_path, '테스트DF.csv')
# 테스트df.to_csv(output_path, encoding='cp949')
# 테스트df

# 시험.입찰명.unique()

# 일종 = categorizeBig(입찰결과물품df, '1종')
# 일종.입찰공고명.unique()
```

```python
def checkDuplicatesInBidName(classA, classB):
    Adf = categorizeBig(입찰결과물품df, classA)
    Bdf = categorizeBig(입찰결과물품df, classB)
    sameDF = Adf[Adf['입찰명'].isin(Bdf['입찰명'])]
    return sameDF
```
