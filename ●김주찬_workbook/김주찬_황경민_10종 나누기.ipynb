```python
### 시험코드

import re
import pandas as pd

def classify_item_advanced(title):
    if not isinstance(title, str):
        return '분류불가(텍스트아님)', '' 

    matched_categories = []
    matched_pick_keywords = set()  # 중복 피하기 위해 set 사용

    for category, rules in category_keywords.items():
        pick_list = rules['pick_list']
        remove_list = rules['remove_list']

        # pick_list의 키워드 중 하나라도 포함하는지 확인
        for pick_keyword in pick_list:  # 개별 키워드로 검사하여 매치된 키워드 수집
            if re.search(re.escape(pick_keyword), title):  # escape로 안전하게 검색
                # remove_list 확인
                if remove_list:
                    remove_pattern = '|'.join(remove_list)
                    if re.search(remove_pattern, title):
                        break  # remove_list에 걸리면 이 카테고리 스킵
                # remove_list에 안 걸리면 카테고리 추가 및 매치된 pick_keyword 수집
                matched_categories.append(category)
                matched_pick_keywords.add(pick_keyword)
                break  # 하나의 키워드라도 매치되면 카테고리 추가 후 다음 카테고리로

    if matched_categories:
        categories_str = ', '.join(sorted(set(matched_categories)))  # 중복 제거하고 정렬
        pick_keywords_str = ', '.join(sorted(matched_pick_keywords))  # 중복 제거하고 정렬
        return categories_str, pick_keywords_str
    else:
        return '미분류', ''

# 4. '입찰명'에 새로운 분류 함수를 적용하여 '카테고리'와 'pick_keywords' 열 추가
if '입찰명' in df.columns:
    df[['카테고리', 'pick_keywords']] = df['입찰명'].apply(lambda x: pd.Series(classify_item_advanced(x)))
    
    # 5. 결과 확인 및 파일 저장
    print("===== 카테고리 분류 결과 (상위 10개) =====")
    print(df[['입찰명', '카테고리', 'pick_keywords']].head(10))

    print("\n===== 카테고리별 개수 요약 =====")
    # 다중 카테고리 때문에 value_counts() 대신 explode로 카운트 (참고용)
    exploded_df = df['카테고리'].str.split(', ').explode()
    print(exploded_df.value_counts())
    
    # 분류 결과를 새로운 CSV 파일로 저장
    output_filename = 'bidResCategorizedUTF.csv'
    df.to_csv(output_filename, index=False, encoding='utf-8')
    
    print(f"\n분류가 완료되었습니다. 결과가 '{output_filename}' 파일로 저장되었습니다.")
    
else:
    print("오류: '입찰명' 열을 찾을 수 없습니다. CSV 파일의 열 이름을 확인해주세요.")
    print("현재 파일의 열 목록:", df.columns.tolist())
```
