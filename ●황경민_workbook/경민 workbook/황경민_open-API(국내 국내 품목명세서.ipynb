```python
def to_extract_row(item, tag_names, col_names=None):
    data = {}
    if col_names is None:
        col_names = tag_names
    for tag, col in zip(tag_names, col_names):
        tagobj = item.find(tag)
        data[col] = tagobj.text.strip() if tagobj else None
    return data
```

```python
PBL_S_df = PBL_df.drop(['참가수', '공고번호', '공고차수', '항목번호', '입찰명', '업무구분', '발주기관', '낙찰자(업체코드)', '낙찰자(상호)', '낙찰자(영문상호)', '낙찰자(대표자)', '낙찰자(연락처)','낙찰자(연락처).1', '낙찰자(주소)', '낙찰률', ' 낙찰금액 '], axis=1)
PBL_S_df.head()
```

```python
rename_dict = {
    '판단번호': 'dcsNo',
    '발주기관코드': 'orntCode',
    '개찰일시': 'demandYear'
}

PBL_S_df = PBL_S_df.rename(columns=rename_dict)

PBL_S_df['demandYear'] = PBL_S_df['demandYear'].astype(str).str[:4]
PBL_S_df.head(8)

PBL_S_df
```

```python
import requests
import pandas as pd
from bs4 import BeautifulSoup
from tqdm import tqdm
import time

base_url = 'http://openapi.d2b.go.kr/openapi/service/BidPblancInfoService/'
service_name = 'getDmstcPrcurePlanDetail'
serviceKey = '8bea1c13542aaa1626e136b275cde504cb37de3c3a31d19cd15d7759b330efdf'

result_list = []
Nodata_list = []

# 'demandYear'가 2016인 데이터만 필터링
pbl_2016_df = PBL_S_df[PBL_S_df['demandYear'] == '2016'].copy()
print(f"✅ 2016년 데이터 {len(pbl_2016_df)}건을 대상으로 테스트를 시작합니다.")

# tqdm 진행률 바 객체를 변수에 저장합니다.
progress_bar = tqdm(pbl_2016_df.iterrows(), total=len(pbl_2016_df), desc="2016년 데이터 호출 중")

# for 루프에서 이 객체를 사용합니다.
for index, row in progress_bar:
    demandYear = row['demandYear']
    orntCode = row['orntCode']
    dcsNo = row['dcsNo']

    params = f'serviceKey={serviceKey}&demandYear={demandYear}&orntCode={orntCode}&dcsNo={dcsNo}'
    url = f'{base_url}{service_name}?{params}'

    try:
        response = requests.get(url, timeout=5)
        soup = BeautifulSoup(response.content, 'xml')
        
        item = soup.find('item')
        if item:
            data_dict = {tag.name: tag.text for tag in item.find_all()}
            result_list.append(data_dict)
        else:
            no_data_info = {'demandYear': demandYear, 'orntCode': orntCode, 'dcsNo': dcsNo}
            Nodata_list.append(no_data_info)

    except Exception as e:
        error_info = {'demandYear': demandYear, 'orntCode': orntCode, 'dcsNo': f"Error: {e}"}
        Nodata_list.append(error_info)
    
    # --- 이 부분이 변경되었습니다 ---
    # 진행률 바의 오른쪽에 표시될 텍스트를 실시간으로 업데이트합니다.
    progress_bar.set_postfix(성공=len(result_list), 실패=len(Nodata_list))
    # -----------------------------

# 모든 결과를 각각의 데이터프레임으로 변환
result_df = pd.DataFrame(result_list)
Nodata_df = pd.DataFrame(Nodata_list)

# 최종 결과 확인
print("\n" + "="*50)
print("\n--- ✅ API 호출 성공 데이터프레임 (2016년) ---")
print(result_df)

print("\n--- ❌ 데이터를 찾지 못한 목록 (2016년) ---")
print(Nodata_df)
```
