```python
base_url = 'http://openapi.d2b.go.kr/openapi/service/BidPblancInfoService/'
service_name = 'getDmstcCmpetBidPblancItem'
serviceKey = '8bea1c13542aaa1626e136b275cde504cb37de3c3a31d19cd15d7759b330efdf'

result_list = []
fail_list = [] 

for index, row in PBL_S_df.iterrows():
    demandYear = row['demandYear']
    orntCode = row['orntCode']
    dcsNo = row['dcsNo']

    params = f'serviceKey={serviceKey}&demandYear={demandYear}&orntCode={orntCode}&dcsNo={dcsNo}'
    url = f'{base_url}{service_name}?{params}'
    
    print(f"{index+1}번째 행 호출 중... URL: {url}")

    try:
        response = requests.get(url, timeout=5)
        soup = BeautifulSoup(response.content, 'xml')
        
        item = soup.find('item')
        if item:
            data_dict = {tag.name: tag.text for tag in item.find_all()}
            fail_list.append(data_dict)
        else:
            print(f"  -> {index+1}번째 행({dcsNo})에서 데이터를 찾지 못했습니다.")
            no_data_info = {
                'demandYear': demandYear,
                'orntCode': orntCode,
                'dcsNo': dcsNo
            }
            fail_list.append(no_data_info)

    except Exception as e:
        print(f"  -> {index+1}번째 행({dcsNo})에서 오류 발생: {e}")

result_df = pd.DataFrame(result_list)
Nodata_df = pd.DataFrame(fail_list)

# 최종 결과 확인
print("\n--- ✅ API 호출 성공 데이터프레임 ---")
print(result_df)

print("\n--- ❌ 데이터를 찾지 못한 목록 ---")
print(Nodata_df)
```

```python
import requests
import pandas as pd
from bs4 import BeautifulSoup
from tqdm import tqdm
import time

base_url = 'http://openapi.d2b.go.kr/openapi/service/BidPblancInfoService/'
service_name = 'getDmstcCmpetBidPblancItem'
serviceKey = '8bea1c13542aaa1626e136b275cde504cb37de3c3a31d19cd15d7759b330efdf'

result_list = []
Nodata_list = []

# 'demandYear'가 2016인 데이터만 필터링
pbl_2016_df = PBL_S_df[PBL_S_df['demandYear'] == '2016'].copy()
print(f"✅ 2016년 데이터 {len(pbl_2016_df)}건을 대상으로 테스트를 시작합니다.")

# tqdm 진행률 바 객체를 변수에 저장합니다.
progress_bar = tqdm(pbl_2016_df.iterrows(), total=len(pbl_2016_df), desc="2016년 데이터 호출 중")

# for 루프에서 이 객체를 사용합니다.
for index, row in progress_bar:
    demandYear = row['demandYear']
    orntCode = row['orntCode']
    dcsNo = row['dcsNo']

    params = f'serviceKey={serviceKey}&demandYear={demandYear}&orntCode={orntCode}&dcsNo={dcsNo}'
    url = f'{base_url}{service_name}?{params}'

    try:
        response = requests.get(url, timeout=5)
        soup = BeautifulSoup(response.content, 'xml')
        
        item = soup.find('item')
        if item:
            data_dict = {tag.name: tag.text for tag in item.find_all()}
            result_list.append(data_dict)
        else:
            no_data_info = {'demandYear': demandYear, 'orntCode': orntCode, 'dcsNo': dcsNo}
            Nodata_list.append(no_data_info)

    except Exception as e:
        error_info = {'demandYear': demandYear, 'orntCode': orntCode, 'dcsNo': f"Error: {e}"}
        Nodata_list.append(error_info)
    
    # --- 이 부분이 변경되었습니다 ---
    # 진행률 바의 오른쪽에 표시될 텍스트를 실시간으로 업데이트합니다.
    progress_bar.set_postfix(성공=len(result_list), 실패=len(Nodata_list))
    # -----------------------------

# 모든 결과를 각각의 데이터프레임으로 변환
result_df = pd.DataFrame(result_list)
Nodata_df = pd.DataFrame(Nodata_list)

# 최종 결과 확인
print("\n" + "="*50)
print("\n--- ✅ API 호출 성공 데이터프레임 (2016년) ---")
print(result_df)

print("\n--- ❌ 데이터를 찾지 못한 목록 (2016년) ---")
print(Nodata_df)
```

```python
import pandas as pd

# 1. CSV 파일 읽기
try:
    df = pd.read_csv('품목명.csv', encoding='cp949')
    # CSV 파일에 '품명'이라는 컬럼이 있는지 확인
    if '품명' not in df.columns:
        print("오류: '품목명.csv' 파일에 '품명' 컬럼이 존재하지 않습니다.")
        # 만약 컬럼명이 다르다면, 여기서 df.rename() 등을 사용하여 '품명'으로 변경해야 합니다.
        exit()
except FileNotFoundError:
    print("오류: '품목명.csv' 파일을 찾을 수 없습니다. 파일이 현재 위치에 있는지 확인해주세요.")
    exit()

# 2. 분류 기준 정의 (키워드 기반)
# 각 종별로 대표적인 키워드를 리스트 형태로 정의합니다. 우선순위가 높은 것을 먼저 배치합니다.
classification_keywords = {
    '5종 탄약류': ['수류탄', '탄약', '지뢰', '폭약', '신관', '포탄', '유도탄'],
    '7종 장비류': ['헬기', '레이더', '무기', '장비', '전차', '장갑차', 'CCTV', '크레인', '탐지기', '보일러', '냉난방기', '작업차'],
    '9종 수리부속·공구류': ['수리부속', '공구', '정비', '조립체', '오-링', '연결축', '기어'],
    '8종 의무 장비·물자': ['의무', '의약품'],
    '3종 유류': ['유류', 'LPG', '가스', '경유', '휘발유', '등유', '윤활유', '냉매', '연료'],
    '4. 종 건설자재': ['목재', '철근', '시멘트', '페인트', '철조망', '철책', '패널', '자재'],
    '1종 식량류': ['식량', '양념', '소스', '식초', '우유', '주스', '김치', '두부', '쌀국수', '시리얼', '돈까스', '소시지', '햄', '빵', '떡', '탕', '전분', '초코파이', '분식', '면류', '건빵'],
    '2종 일반물자류': ['명찰', '도시락', '분전반', '가구', '비품', '조명', '피복', '침구', '장구', '기재', '기구', '전지', '무전기', '전산', '소화기', '지도', '반합', '책상', '펜', '배터리'],
    '6종 복지매장 판매품(PX 물품)': ['PX'] # PX는 키워드가 명확하지 않아, 식품류/일반물자류와 겹치는 경우가 많습니다.
}

# 3. 품목명을 분류하는 함수 정의
def classify_item(item_name):
    """품목명을 입력받아 해당하는 군수품 종별을 반환하는 함수"""
    # item_name이 문자열이 아닐 경우를 대비
    if not isinstance(item_name, str):
        return '10종 기타'
    
    # 정의된 키워드 딕셔너리를 순회하며 검사
    for category, keywords in classification_keywords.items():
        for keyword in keywords:
            if keyword in item_name:
                return category # 키워드가 발견되면 해당 종별을 반환
    
    # 모든 키워드에 해당하지 않으면 '10종 기타'로 분류
    return '10종 기타'

# 4. '품명' 컬럼에 분류 함수를 적용하여 '분류' 컬럼 생성
df['분류'] = df['품명'].apply(classify_item)

# 5. 결과 확인
df
```
