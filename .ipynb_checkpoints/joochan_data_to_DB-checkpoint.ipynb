```python
import pymysql
pymysql.__version__
```

```python
# DB 연결 객체 생성
conn = pymysql.connect(host='127.0.0.1', user='root', password='1234', charset='utf8mb4')
conn
```

```python
# 커서 객체 생성
cur = conn.cursor()

# DB 생성
sql = 'create database if not exists bidRes_itemsDB default character set utf8mb4;'
cur.execute(sql)
```

```python
cur.execute('show databases;')
```
