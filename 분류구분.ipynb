```python
import requests
import pandas as pd
import math

base_url     = 'https://api.odcloud.kr/api/'
service_name = '15050917/v1/uddi:242ab8e6-baef-41f0-a495-b6994aff720e_201812101026'
perPage      = 1000   
returnType   = 'json'
serviceKey   = '416a0c4c87da58598c1fde059d56608bca0746c6d66fffb18def03dee40491a6'

all_rows = []

url = f"{base_url}{service_name}?serviceKey={serviceKey}&page=1&perPage={perPage}&returnType={returnType}"
res = requests.get(url)
if res.status_code == 200:
    data = res.json()
    totalCount = data.get("totalCount", 0)
    totalPages = math.ceil(totalCount / perPage)
    print(f"총 데이터 건수: {totalCount}, 총 페이지 수: {totalPages}")
    all_rows.extend(data.get("data", []))
else:
    raise Exception(f"❌ 에러: {res.status_code} {res.text}")

for page_num in range(2, totalPages+1):
    url = f"{base_url}{service_name}?serviceKey={serviceKey}&page={page_num}&perPage={perPage}&returnType={returnType}"
    res = requests.get(url)
    if res.status_code == 200:
        data = res.json()
        all_rows.extend(data.get("data", []))
    else:
        print("❌ 에러:", res.status_code, res.text)
        break

df = pd.DataFrame(all_rows)
```

```python
import pandas as pd

# CSV 불러오기
df = pd.read_csv("수요기관_전체분류.csv", encoding="utf-8-sig")

keywords_2jong = {
    "피복류": ["군복", "피복", "방한", "침구", "장구", "의류", "방탄복", "군화", "모자", "양말", "티셔츠", "속옷"],
    "병참물자": ["비품", "기재", "취사", "냉난방", "군악", "체육", "교육", "포장재", "로프", "안전용품"],
    "화생방물자": ["방독면", "제독제", "연막", "오염표지판", "탐지기", "보호장비"],
    "통신물자": ["전지", "배터리", "야전선", "CCTV", "무전기", "전화", "라디오", "위성TV", "네트워크", "전산", "컴퓨터"],
    "공병물자": ["소방", "소화기", "위장망"],
    "기타": ["군사지도", "드론"]
}

def classify_2jong(text):
    """ 2종 """
    if pd.isna(text):
        return None
    for category, kws in keywords_2jong.items():
        for kw in kws:
            if kw in str(text):
                return f"2종-{category}"
    return None

df["군수품분류"] = df["입찰공고명"].apply(classify_2jong)

print(df[["입찰공고명", "군수품분류"]].head(20))
```

```python
keywords_3jong = {
    "일반유류": ["경유", "휘발유", "등유"],
    "항공류": ["제트유", "항공유", "항공휘발유", "AV-GAS"],
    "윤활유": ["엔진오일", "기어오일", "유압작동오일", "압축기오일", "그리스", "방청유"],
    "화공약품류": ["방청제", "세관제", "솔벤트", "부동액", "제초제", "정수약품", "살충제"],
    "고체연료": ["목재펠릿", "연탄", "조개탄", "착화탄"],
    "가스류": ["LPG", "LNG", "CNG", "아세틸렌", "냉매", "가스"],
    "유류포장재료": ["드럼", "가스용기"]
}

def classify_3jong(text):
    """ 3종(유류) """
    if pd.isna(text):
        return None
    for category, kws in keywords_3jong.items():
        for kw in kws:
            if kw in str(text):
                return f"3종-{category}"
    return None

df["군수품분류"] = df["군수품분류"].combine_first(
    df["입찰공고명"].apply(classify_3jong)
)

df[df["군수품분류"].notna()][["입찰공고명", "군수품분류"]].head(20)
```

```python
keywords_5jong = {
    "지상탄약": ["소구경탄", "직사화기탄", "박격포탄", "포병탄", "지대지유도탄", "수류탄", "지뢰", "연막탄", "신호탄", "화학탄", "폭약"],
    "해상탄약": ["함포탄", "기뢰", "폭뢰", "함정유도탄"],
    "항공탄약": ["항공유도탄", "항공폭탄", "투하탄", "일반폭탄", "확산탄", "기관포탄", "항공로켓탄", "자폭드론"],
    "기타": ["탄피", "탄약포장", "폭파기재"]
}

def classify_5jong(text):
    """ 5종(탄약류) """
    if pd.isna(text):
        return None
    for category, kws in keywords_5jong.items():
        for kw in kws:
            if kw in str(text):
                return f"5종-{category}"
    return None

df["군수품분류"] = df["군수품분류"].combine_first(
    df["입찰공고명"].apply(classify_5jong)
)

df[df["군수품분류"].notna()][["입찰공고명", "군수품분류"]].head(20)
```

```python
# 2종 키워드
keywords_2jong = {
    "통신물자": ["전산운영용품", "보조배터리", "전산시스템", "유지보수", "통신장비"],
    "병참물자": ["교육용", "영상교재", "훈련", "체육기구", "취사", "냉난방", "재보급"],
    "피복류": ["보장구", "피복", "방진화", "티셔츠"],
    "화생방물자": ["누설탐지기", "보호의", "방독면"]
}

# 3종 키워드
keywords_3jong = {
    "가스류": ["LPG", "액화석유가스", "의료용가스", "가스터빈엔진"],
    "일반유류": ["휘발유", "경유", "유류"],
    "유류포장재료": ["드럼"]
}

# 5종 키워드
keywords_5jong = {
    "지상탄약": ["지뢰제거", "탄약", "수류탄", "포탄", "폭약", "신호탄", "연막탄"]
}

def classify_2_3_5(text):
    """입찰공고명에서 2·3·5종 키워드 찾기"""
    if pd.isna(text):
        return None
    s = str(text)
    
    # 2종
    for category, kws in keywords_2jong.items():
        for kw in kws:
            if kw in s:
                return f"2종-{category}"
    
    # 3종
    for category, kws in keywords_3jong.items():
        for kw in kws:
            if kw in s:
                return f"3종-{category}"
    
    # 5종
    for category, kws in keywords_5jong.items():
        for kw in kws:
            if kw in s:
                return f"5종-{category}"
    
    return None

df["군수품분류"] = df["입찰공고명"].apply(classify_2_3_5)

df[df["군수품분류"].notna()][["입찰공고명", "군수품분류"]].head(30)
```
